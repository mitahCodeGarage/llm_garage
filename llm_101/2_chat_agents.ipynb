# Chat Agents
This is a sample code of 2 chatbots which are communicating with each other. Their personas are defined below

## Personas
- Customer : Who wishes to buy a paracetomol and has crocin as preference.
- Sales Agent : Who is at the pharmacy and eager to upsell products to the customers

## Scenario
Customer wants to buy crocin, but crocin is out of stock in the pharmacy. The Sales Agent then suggests Dolo against the crocin

```python
import os
from litellm import completion
from dotenv import load_dotenv
```

```python
load_dotenv()
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")
```

```python
conv_dialogue = ["I want to buy Crocin"]

system_prompt_seller = """You are chatbot Sales agent of a Pharmacy shop. 
You always try to sell additional stuff to the customers apart from their requirement. 
Also incase a medicine is not available, you try to persuade the customer to buy some other medicine. 
You have Dolo as a paracetomol medicine and crocin is out of stock.
You are in a conversation with a customer
Only give the response and nothing else"""

user_prompt_seller = f"""The conversation so far is as follows
{conv_dialogue}
Now with this, respond with what ou would like to say next, as a sales agent"""

system_prompt_buyer = """You are a chatbot acting as a customer who wants to purchase a paracetomol. 
You prefer crocin and want to buy crocin. But as you are in pain are ready to look for other options. You always enquire before closing on the purchase.
You are in a conversation with the sales agent
Only give the response and nothing else"""

user_prompt_buyer = f"""The conversation so far is as follows
{conv_dialogue}
Now with this, respond with what ou would like to say next, as a customer"""
```

```python
def customer_dialogue():
    
    user_prompt_buyer = f"""The conversation so far is as follows
    {conv_dialogue}
    Now with this, respond with what you would like to say next, as a customer"""

    message = [
        {
            "role" : "system",
            "content" : system_prompt_buyer
        },
        {
            "role" : "user",
            "content" : user_prompt_buyer
        }
    ]
    #print(message)
    response = completion(base_url="http://localhost:11434", model="ollama/qwen3:8b", messages = message)
    #response = completion(base_url="https://openrouter.ai/api/v1", api_key=openrouter_api_key, model="openrouter/google/gemini-2.5-flash", messages=message)

    print(response.choices[0].message.content)
    return response.choices[0].message.content
```

```python
def sales_agent_dialogue():
    user_prompt_seller = f"""The conversation so far is as follows
    {conv_dialogue}
    Now with this, respond with what you would like to say next, as a sales agent"""
    
    message = [
        {
            "role" : "system",
            "content" : system_prompt_seller
        },
        {
            "role" : "user",
            "content" : user_prompt_seller
        }
    ]
    #print(message)
    response = completion(base_url="http://localhost:11434", model="ollama/qwen3:8b", messages = message)
    #response = completion(base_url="https://openrouter.ai/api/v1", api_key=openrouter_api_key, model="openrouter/google/gemini-2.5-flash", messages=message)

    print(response.choices[0].message.content)
    return response.choices[0].message.content
```
