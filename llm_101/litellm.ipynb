# This code snippet is to show how to use LiteLLM

```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown
from litellm import completion
```

```python
load_dotenv()
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")

if openrouter_api_key:
    print("Open router key is configured")
else:
    print("Open router key is missing")
```

```python
message = [
    {
        "role": "user",
        "content" : "Tell a joke for a student on the journey to becoming an expert in LLM Engineering"
    }
]
```

## Using Gemini via Openrouter

```python
response = completion(base_url="https://openrouter.ai/api/v1", api_key=openrouter_api_key, model="openrouter/google/gemini-2.5-flash", messages=message)
display(Markdown(response.choices[0].message.content))
```

## Using Qwen3 via Ollama in local

```python
response = completion(base_url="http://localhost:11434", model ="ollama/qwen3:8b", api_key="ollama", messages = message)
display(Markdown(response.choices[0].message.content))
```

```python
print(f"Input Tokens : {response.usage.prompt_tokens}")
print(f"Output Tokens : {response.usage.completion_tokens}")
print(f"Total Tokens : {response.usage.total_tokens}")
print(f"Output Tokens : {response.usage}")
```
