# This code snippet is to show how to use Langchain

```python
import os
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from IPython.display import Markdown
```

```python
load_dotenv()
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")

if openrouter_api_key:
    print("Key is configured")
else:
    print("Key is not configured")
```

```python
tell_a_joke = [
    {
        "role":"user",
        "content" : "Tell a joke for a student on the journey to becoming an expert in LLM Engineering"
    }
]
```

## Using GPT model via Openrouter

```python
llm = ChatOpenAI(api_key=openrouter_api_key, model="openai/gpt-6-luna", reasoning_effort="none", base_url="https://openrouter.ai/api/v1",)
response = llm.invoke(tell_a_joke)

display(Markdown(response.content))
```

## Using Qwen3 model via Ollama on local

```python
llm = ChatOpenAI(api_key=openrouter_api_key, model="qwen3:8b", reasoning_effort="none", base_url="http://localhost:11434/v1")
response = llm.invoke(tell_a_joke)

display(Markdown(response.content))
```
