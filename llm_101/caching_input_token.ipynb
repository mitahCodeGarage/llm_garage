# Caching of input token
This needs the input prompt to have 1024 tokens atleast

```python
import os
from dotenv import load_dotenv
from IPython.display import Markdown
from litellm import completion
```

```python
load_dotenv()
openrouter_api_key = os.getenv("OPENROUTER_API_KEY")

if openrouter_api_key:
    print("Open router key is configured")
else:
    print("Open router key is missing")
```

```python
base_context = """
LangChain is a framework for building applications powered by
large language models (LLMs). It supports prompt templates,
model integrations, chains, tools, retrieval-augmented generation
(RAG), and AI agents. OpenRouter provides a unified API to access
models from multiple providers, including Google Gemini. My name is Hatim.
""" * 30

question = """
What is my name?"""
```

```python
message = [
    {
        "role": "system",
        "content" : base_context
    },
    {
        "role": "user",
        "content" : question
    }
]
```

```python
response = completion(base_url="https://openrouter.ai/api/v1", api_key=openrouter_api_key, model="openrouter/google/gemini-2.5-flash", messages=message)
display(Markdown(response.choices[0].message.content))
```

```python
print(f"Input Token : {response.usage.prompt_tokens}")
print(f"Cached Tokens : {response.usage.prompt_tokens_details.cached_tokens}")
print(f"Output Tokens : {response.usage}")
```
